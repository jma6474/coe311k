```python
import numpy as np 
```

```python
v1 = np.array([1,2,3])
v2 = np.array([4,5,6])
print (v1)
print (v2)
print (v1+v2)
print (v1*v2)
dotProduct = np.dot(v1,v2)
print (dotProduct)
magnitude = np.linalg.norm(v1) 
print (magnitude) #magnitude also called norm 
```

```python
matrix = np.array([[1,2,3],[4,5,6]])
print(matrix)
matrix2 = np.array([[7,8,9],[10,11,12]])
print()
print(matrix2)
print()
print(matrix - matrix2)
print()
#transpose.matrix.T
```

```python
#2x+3y = 5       4x-y =3
#([[2,3],[4,-1]])([[x],[y]])=([[5],[-3])
A = np.array([[2,3],[4,-1]])
b = np.array([5,3])
x = np.linalg.solve(A,b)
print(x)
```

```python
#eigenvalue stuff 

A = np.array([[2,1],[1,2]])
e_value, e_vectors = np.linalg.eig(A)   #eig returns two things 
print("eiganvalue: ", e_value)
print("eiganvector: ", e_vectors)
print()

# inverse of a matrix 

A = np.array([[4,7],[2,6]])
A_inverse = np.linalg.inv(A)
print(A_inverse)
print()

identity = np.dot(A, A_inverse)
print(identity)
# 1 0 0
# 0 1 0
# 0 0 1
```

```python
#singular value decomposition 

A = np.array([[1,2],[3,4],[5,6]])
U, sigma, Vt = np.linalg.svd(A)
print(U)
print()
print(sigma)
print()
print(Vt)
```
