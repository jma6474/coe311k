```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.linspace(0, 2 * np.pi, 50)
y = np.random.rand(1000)

plt.subplot(2, 1, 2)
plt.plot(x, np.cos(x), 'g')
plt.show()

plt.subplot(1, 1, 1)
plt.hist(y, 50)
plt.show()

plt.subplot(2, 2, 2)
y = np.sin(x)
plt.scatter(x,y)

data = [23, 17, 35, 29, 12, 41]
fig = plt.figure(figsize=(10, 7))
plt.pie(data)
plt.show()
```
