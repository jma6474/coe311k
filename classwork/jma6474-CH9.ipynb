```python
import numpy as np
import matplotlib.pyplot as plt

# Given data
X = np.array([1, 2, 3, 4, 5])
Y = np.array([1.2, 2.8, 3.6, 4.5, 5.1])

# Least Squares
A = np.vstack([X, np.ones(len(X))]).T
m, c = np.linalg.lstsq(A, Y, rcond=None)[0]

print("Part 1")
print("Slope:", m)
print("Intercept:", c)

# Plot
t = np.linspace(min(X), max(X), 100)
plt.scatter(X, Y)
plt.plot(t, m*t + c)
plt.title("Least Squares - Given Data")
plt.show()
```

```python
# Randomize data
X = np.random.uniform(1, 10, 5)
Y = np.random.uniform(1, 10, 5)

A = np.vstack([X, np.ones(len(X))]).T
m, c = np.linalg.lstsq(A, Y, rcond=None)[0]

print("Part 2")
print("Slope:", m)
print("Intercept:", c)

t = np.linspace(min(X), max(X), 100)
plt.scatter(X, Y)
plt.plot(t, m*t + c)
plt.title("Least Squares - Random Data")
plt.show()
```

```python
# True equation
m_true = 3
b_true = 2

X = np.arange(0, 1.01, 0.01)

# Add symmetric noise
Y = m_true*X + b_true + np.random.uniform(-0.5, 0.5, len(X))

# Least Squares
A = np.vstack([X, np.ones(len(X))]).T
m, c = np.linalg.lstsq(A, Y, rcond=None)[0]

print("Part 3")
print("True slope:", m_true)
print("Fitted slope:", m)
print("True intercept:", b_true)
print("Fitted intercept:", c)

plt.scatter(X, Y)
plt.plot(X, m*X + c)
plt.title("Least Squares - Noisy Linear Data")
plt.show()
```
