```python
import numpy as np
import matplotlib.pyplot as plt

x_start, x_end = 0, 2

def f(x, y):
    return -2 * y

for run_num, h in enumerate(np.arange(0.01, 1.01, 0.05)):

    h = round(h, 2)
    num_steps = int((x_end - x_start) / h)
    x_values = np.linspace(x_start, x_end, num_steps + 1)
    y_values = np.zeros(num_steps + 1)
    y_values[0] = 1

    for i in range(num_steps):
        y_values[i + 1] = y_values[i] + h * f(x_values[i], y_values[i])

    if run_num == 0:
        first_run_x = x_values
        first_run_y = y_values

    print("Step size h =", h)
    total_error = 0
    for i in range(len(x_values)):
        ref = np.interp(x_values[i], first_run_x, first_run_y)
        error = abs(y_values[i] - ref)
        total_error = total_error + error
        print("  x =", round(x_values[i], 4), "  error =", round(error, 8))
    print("  Total error =", round(total_error, 6))
    print()

    plt.plot(x_values, y_values, label="h=" + str(h))

plt.xlabel('x')
plt.ylabel('y')
plt.title("Eulers Method")
plt.legend()
plt.show()
```
