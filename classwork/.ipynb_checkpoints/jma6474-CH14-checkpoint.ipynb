```python
import numpy as np
import matplotlib.pyplot as plt

def f(x):
    return np.sin(x)

a = 0
b = 2 * np.pi
N10 = 10
N50 = 50

x10 = np.linspace(a, b, N10)
x50 = np.linspace(a, b, N50)

h10 = (b - a) / (N10 - 1)
h50 = (b - a) / (N50 - 1)

# True derivative
true_deriv10 = np.cos(x10)
true_deriv50 = np.cos(x50)

# Central Difference Approximation
central10 = (f(x10 + h10) - f(x10 - h10)) / (2 * h10)
central50 = (f(x50 + h50) - f(x50 - h50)) / (2 * h50)

# Forward Difference Approximation
forward10 = (f(x10 + h10) - f(x10)) / h10
forward50 = (f(x50 + h50) - f(x50)) / h50

# Backward Difference Approximation
backward10 = (f(x10) - f(x10 - h10)) / h10
backward50 = (f(x50) - f(x50 - h50)) / h50

for x, true_deriv, central, forward, backward, N in [
    (x10, true_deriv10, central10, forward10, backward10, 10),
    (x50, true_deriv50, central50, forward50, backward50, 50)
]:
    fig, axes = plt.subplots(2, 2, figsize=(10, 7))
    fig.suptitle(f'N = {N}')

    axes[0, 0].plot(x, true_deriv, color='black')
    axes[0, 0].set_title('cos(x) — True Derivative')

    axes[0, 1].plot(x, forward, color='blue')
    axes[0, 1].set_title('Forward Difference')

    axes[1, 0].plot(x, backward, color='gray')
    axes[1, 0].set_title('Backward Difference')

    axes[1, 1].plot(x, central, color='red')
    axes[1, 1].set_title('Central Difference')

    plt.tight_layout()
    plt.show()
```
