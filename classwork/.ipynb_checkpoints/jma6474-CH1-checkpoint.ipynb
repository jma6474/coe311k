```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 2*np.pi, 50)
y = np.random.rand(1000)

plt.figure(figsize=(10, 8))

plt.subplot(2, 2, 1)
plt.plot(x, np.cos(x))

plt.subplot(2, 2, 2)
plt.hist(y, bins=50)

plt.subplot(2, 2, 3)
plt.scatter(x, np.sin(x))

plt.subplot(2, 2, 4)
plt.pie([23, 17, 35, 29, 12, 41])

plt.tight_layout()
plt.show()
```
