```python
import random
import numpy as np 
import math

# Monte Carlo method to approximate pi
def approxPI(numSamples):
    counter = 0

    for i in range(numSamples):
        x = random.uniform(-1, 1)
        y = random.uniform(-1, 1)

        if x*x + y*y < 1:
            counter += 1

    return 4 * counter / numSamples

for i in range (0,100):
    calculatedPI = approxPI(1000)
    truePI = np.pi
    trueERROR = abs(truePI-calculatedPI)
    relativeERROR = abs((truePI - calculatedPI)/truePI)
    print(calculatedPI, trueERROR, relativeERROR)
```
