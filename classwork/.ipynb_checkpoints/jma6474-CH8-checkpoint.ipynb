```python
import numpy as np
import matplotlib.pyplot as plt

X = np.random.uniform(1, 10, 5)
Y = np.random.uniform(1, 10, 5)

X.sort()

A = np.vstack([X, np.ones(len(X))]).T
m, c = np.linalg.lstsq(A, Y, rcond=None)[0]
print(f"Slope: {m}, Intercept: {c}")

t = np.linspace(np.min(X), np.max(X), 100)
y = m*t + c

plt.scatter(X, Y, color='blue', label='Random Data')
plt.plot(t, y, color='red', label='Least Squares Fit')
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Least Squares')
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

m_true = np.random.rand()*5
b_true = np.random.rand()*5

X = np.arange(0, 1.01, 0.01)
X.sort()

Y = m_true*X + b_true + np.random.rand(len(X))

A = np.vstack([X, np.ones(len(X))]).T
m, c = np.linalg.lstsq(A, Y, rcond=None)[0]

print("True slope:", m_true)
print("Fitted slope:", m)
print(f"True intercept:", b_true)
print(f"Fitted intercept:", c)

y = m*X + c
residuals = Y - y
max = np.argmax(residuals)
min = np.argmin(residuals)
print("Largest residual:", residuals[max], "at X =", X[max])
print("Smallest residual:", residuals[min], "at X =", X[min])

# Plot
plt.scatter(X, Y, color='blue', label='Data points')
plt.plot(X, y, color='red', label='Least Squares Fit')
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Least Squares Fit with error')
plt.legend()
plt.show()
```
