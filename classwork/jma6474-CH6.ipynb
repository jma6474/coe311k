```python
import numpy as np
import time
# gauss seidel

def gauss_seidel(A, b, x0, tol, max_iterations):
    n = len(b)
    x = x0.copy()
    for k in range(max_iterations):
        x_old = x.copy()
        
        for i in range(n):
            s1 = sum(A[i][j] * x[j] for j in range(i))
            s2 = sum(A[i][j] * x_old[j] for j in range(i + 1, n))
            x[i] = (b[i] - s1 - s2) / A[i][i]
        
        # Check for convergence after full iteration
        if np.linalg.norm(x - x_old, ord=np.inf) < tol:
            return x, k + 1
    return x, max_iterations

# jacobi
def jacobi(A, b, x0, tol, max_iterations):
    n = len(b)
    x = x0.copy()
    for k in range(max_iterations):
        x_new = np.zeros_like(x)
        for i in range(n):
            s = sum(A[i][j] * x[j] for j in range(n) if j != i)
            x_new[i] = (b[i] - s) / A[i][i]
        # Check for convergence
        if np.linalg.norm(x_new - x, ord=np.inf) < tol:
            return x_new, k + 1
        x = x_new
    return x, max_iterations

# Example usage
start_time = time.perf_counter()

A = np.array([[10, -1, 2, 0, 0],
              [-1, 11, -1, 3, 0],
              [2, -1, 10, -1, 0],
              [0, 3, -1, 8, -2],
              [0, 0, 0, -2, 9]])
b = np.array([14, 30, 26, 25, 37])
x0 = np.zeros_like(b)
tol = 1e-6
max_iterations = 100

solutionJACOBI, iterationsJACOBI = jacobi(A, b, x0, tol, max_iterations)
solutionGAUSS, iterationsGAUSS = gauss_seidel(A, b, x0, tol, max_iterations)

end_time = time.perf_counter()
elapsed_time = end_time - start_time

print("Solutions: Jacobi- ", solutionJACOBI, "Gauss seidel- ", solutionGAUSS)
print("Iterations: Jacobi- ", iterationsJACOBI, "Gauss seidel- ", iterationsGAUSS)
print("Time: ", elapsed_time)
```
