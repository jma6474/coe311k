```python
import random
import numpy as np
import matplotlib.pyplot as plt

def approxPI(numSamples):
    counter = 0
    for _ in range(numSamples):
        x = random.uniform(-1, 1)
        y = random.uniform(-1, 1)
        if x*x + y*y < 1:
            counter += 1
    return 4 * counter / numSamples

numTrials = 100
calculatedPI = np.array([approxPI(1000) for _ in range(numTrials)])
truePI = np.pi

true_errors = np.abs(truePI - calculatedPI)
relative_errors = true_errors / truePI

plt.plot(true_errors, label="True Error")
plt.plot(relative_errors, label="Relative Error")
plt.xlabel("Trial")
plt.ylabel("Error")
plt.title("Monte Carlo π Approximation Errors")
plt.legend()
plt.show()
```
