```python
import numpy as np 
import time
import matplotlib.pyplot as plt

# matrix multiplier function
def naive_matrix_multiplication(A, B):
    n = len(A)
    C = [[0] * n for _ in range(n)]
    for i in range(n):
        for j in range(n):
            for k in range(n):
                C[i][j] += A[i][k] * B[k][j]
    return C

# matrices     
a1=np.random.randint(0, 11, size=(3, 3))
a2=np.random.randint(0, 11, size=(3, 3))
b1=np.random.randint(0, 11, size=(6, 6))
b2=np.random.randint(0, 11, size=(6, 6))
c1=np.random.randint(0, 11, size=(12, 12))
c2=np.random.randint(0, 11, size=(12, 12))
d1=np.random.randint(0, 11, size=(24, 24))
d2=np.random.randint(0, 11, size=(24, 24))
e1=np.random.randint(0, 11, size=(48, 48))
e2=np.random.randint(0, 11, size=(48, 48))
f1=np.random.randint(0, 11, size=(96, 96))
f2=np.random.randint(0, 11, size=(96, 96))
g1=np.random.randint(0, 11, size=(192, 192))
g2=np.random.randint(0, 11, size=(192, 192))
h1=np.random.randint(0, 11, size=(384, 384))
h2=np.random.randint(0, 11, size=(384, 384))

# numpy matrix multiplier
t = np.zeros(8)
start = time.perf_counter(); np.matmul(a1, a2); t[0] = time.perf_counter() - start
start = time.perf_counter(); np.matmul(b1, b2); t[1] = time.perf_counter() - start
start = time.perf_counter(); np.matmul(c1, c2); t[2] = time.perf_counter() - start
start = time.perf_counter(); np.matmul(d1, d2); t[3] = time.perf_counter() - start
start = time.perf_counter(); np.matmul(e1, e2); t[4] = time.perf_counter() - start
start = time.perf_counter(); np.matmul(f1, f2); t[5] = time.perf_counter() - start
start = time.perf_counter(); np.matmul(g1, g2); t[6] = time.perf_counter() - start
start = time.perf_counter(); np.matmul(h1, h2); t[7] = time.perf_counter() - start
print(t)

# naive matrix multiplier
t2 = np.zeros(8)                                          # ← t2 not t
start = time.perf_counter(); naive_matrix_multiplication(a1.tolist(), a2.tolist()); t2[0] = time.perf_counter() - start
start = time.perf_counter(); naive_matrix_multiplication(b1.tolist(), b2.tolist()); t2[1] = time.perf_counter() - start
start = time.perf_counter(); naive_matrix_multiplication(c1.tolist(), c2.tolist()); t2[2] = time.perf_counter() - start
start = time.perf_counter(); naive_matrix_multiplication(d1.tolist(), d2.tolist()); t2[3] = time.perf_counter() - start
start = time.perf_counter(); naive_matrix_multiplication(e1.tolist(), e2.tolist()); t2[4] = time.perf_counter() - start
start = time.perf_counter(); naive_matrix_multiplication(f1.tolist(), f2.tolist()); t2[5] = time.perf_counter() - start
start = time.perf_counter(); naive_matrix_multiplication(g1.tolist(), g2.tolist()); t2[6] = time.perf_counter() - start
start = time.perf_counter(); naive_matrix_multiplication(h1.tolist(), h2.tolist()); t2[7] = time.perf_counter() - start
print(t2)

# plot
matrix_numbers = [1, 2, 3, 4, 5, 6, 7, 8]
labels = ['3x3','6x6','12x12','24x24','48x48','96x96','192x192','384x384']

plt.figure()
plt.plot(matrix_numbers, t2, 'o-', color='red', linewidth=2, label='Naive')
plt.plot(matrix_numbers, t,  's-', color='blue', linewidth=2, label='NumPy')
plt.xlabel('Matrix Number')
plt.ylabel('Runtime (seconds)')
plt.title('Naive vs NumPy Matrix Multiplication Runtime')
plt.legend()
plt.tight_layout()
plt.show()
```
