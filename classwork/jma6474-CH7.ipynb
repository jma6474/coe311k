```python
import numpy as np
import matplotlib.pyplot as plt

# Gauss-Seidel
def gauss_seidel(A, b, x0, tol, max_iterations):
    n = len(b)
    x = x0.copy()
    errors = []

    for k in range(max_iterations):
        x_old = x.copy()

        for i in range(n):
            s1 = sum(A[i][j] * x[j] for j in range(i))
            s2 = sum(A[i][j] * x_old[j] for j in range(i + 1, n))
            x[i] = (b[i] - s1 - s2) / A[i][i]

        error = np.linalg.norm(x - x_old, ord=np.inf)
        errors.append(error)

        if error < tol:
            return x, k+1, errors

    return x, max_iterations, errors


# Jacobi
def jacobi(A, b, x0, tol, max_iterations):
    n = len(b)

    x = x0.copy()
    errors = []

    for k in range(max_iterations):
        x_new = np.zeros_like(x)

        for i in range(n):
            s = sum(A[i][j] * x[j] for j in range(n) if j != i)
            x_new[i] = (b[i] - s) / A[i][i]

        error = np.linalg.norm(x_new - x, ord=np.inf)
        errors.append(error)

        if error < tol:
            return x_new, k+1, errors

        x = x_new

    return x, max_iterations, errors
```

```python
#part A 
A = np.array([
	[0.582745, 0.48	, 0.10	, 0.  	, 0.  	], 
 [0.48	, 1.044129, 0.46	, 0.10	, 0.  	],
 [0.10	, 0.46	, 1.10431 , 0.44	, 0.10	],
 [0.  	, 0.10	, 0.44	, 0.963889, 0.42	],
 [0.  	, 0.  	, 0.10	, 0.42	, 0.522565]], dtype=float)
b = np.array([1.162745, 2.084129, 2.20431 , 1.923889, 1.042565], dtype=float)
tol = 1e-6
max_iterations = 100
x0 = np.zeros_like(b)

solutionJACOBI, iterationsJACOBI, errorJACOBI = jacobi(A, b, x0, tol, max_iterations)
solutionGAUSS, iterationsGAUSS, errorGAUSS = gauss_seidel(A, b, x0, tol, max_iterations)

print("Solutions: Jacobi- ", solutionJACOBI, "Gauss-Seidel- ", solutionGAUSS)
print("Iterations: Jacobi- ", iterationsJACOBI, "Gauss-Seidel- ", iterationsGAUSS)
```
