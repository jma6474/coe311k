```python
import random

# Monte Carlo method to approximate pi
def approxPI(numSamples):
    counter = 0

    for i in range(numSamples):
        x = random.uniform(-1, 1)
        y = random.uniform(-1, 1)

        if x*x + y*y < 1:
            counter += 1

    return 4 * counter / numSamples

print(approxPI(100000))
```
