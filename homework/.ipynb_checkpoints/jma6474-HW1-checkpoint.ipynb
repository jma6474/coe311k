```python
import numpy as np 
import math 
import matplotlib.pyplot as plt 
import random

def estimatingPI(numSamples):
    insideCircle = 0

    for i in range(numSamples):
        x = random.uniform(-1, 1)
        y = random.uniform(-1, 1)
        if x**2 + y**2 <= 1:
            insideCircle += 1
    return 4 * insideCircle / numSamples

samples = []
errors = []

for k in range(10, 1000, 20):
    estPI = estimatingPI(k)
    error = abs(math.pi - estPI)
    samples.append(k)
    errors.append(error)
plt.figure()
plt.plot(samples, errors)
plt.xlabel("Samples")
plt.ylabel("Error")
plt.show()
```
